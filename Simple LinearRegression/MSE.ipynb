```python
X = [1, 2, 3, 4, 5]
y = [3, 5, 7, 9, 11]
w,b=2,1
y_pred = []
for i in X:
    pred = w*i+b
    y_pred.append(pred)
print(y_pred)
```

**TOTAL**

```python
lst = []
for i in range(len(y)):
    element = (y[i] -y_pred[i])**2
    lst.append(element)
print(lst)
```

*MSE*

```python
total = 0
for i in lst:
    total += i
print(total)
mse = 1/len(lst)*total
print(mse)
```
