```python
X = [1, 2, 3, 4, 5]
y = [3, 5, 7, 9, 11]
w,b=2,1
y_pred = []
for i in X:
    pred = w*i + b
    y_pred.append(pred)
print(y_pred)
```

*TOTAL*

```python
lst = []
for i in range(len(y)):
    element = abs(y[i]-y_pred[i])
    lst.append(element)
print(lst)
```

*CALCULATE MAE*

```python
total = 0
for i in lst:
    total += i
print(total)
mae = 1/len(X)*total
print(mae)
```

**MAE=0**
