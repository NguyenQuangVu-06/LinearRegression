```python

X = [1, 2, 3, 4, 5]
y = [3, 5, 7, 9, 11]
w,b=2,1
y_pred = []
for i in X:
    pred = w*i+b
    y_pred.append(pred)
print(y_pred)
```

*TOTAL*

```python
lst = []
for i in range(len(y)):
    element = (y[i] -y_pred[i])**2
    lst.append(element)
print(lst)
```

*CALCULATE RMSE* #
*RMSE IS THE SQUARE ROOT OF MSE*

```python
import math
total = 0
for i in lst:
    total += i
print(total)
mse = 1/len(lst)*total
rmse = math.sqrt(mse)
print(rmse)
```
