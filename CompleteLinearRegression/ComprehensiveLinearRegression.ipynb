```python
import math
X = [1, 2, 3, 4, 5, 6, 7, 8]
y = [3.2, 4.8, 7.1, 8.5,11.3, 12.4, 15.2, 16.1]
w,b=0,0
learning_rate = 0.01
lst = []
epochs = 1000
for e in range(epochs):
    g_w , g_b = 0 , 0
    for i in range(len(X)):
        pred = w*X[i] + b
        g_w += (pred - y[i])*X[i]
        g_b += (pred - y[i])
    g_w = 2*g_w/len(X)
    g_b = 2*g_b/len(X)
    w = w - learning_rate * g_w
    b = b - learning_rate * g_b
print("W:",w)
print("B:",b)
y_pred = []
for i in range(len(X)):
    pred = w*X[i] + b
    y_pred.append(pred)
mse,mae = 0,0
for i in range(len(X)):
    mse += (y[i] -y_pred[i])**2
    mae += abs(y[i] -y_pred[i])
mse = mse/len(X)
mae = mae/len(X)
rmse = mse**0.5
print("MSE:",mse)
print("MAE:",mae)
print("RMSE:",rmse)
```
