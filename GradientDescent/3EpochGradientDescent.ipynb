**The old example is used 1 epoch , but now will use to 3 epochs**

```python
X = [1, 2, 3]
y = [3, 5, 7]
w,b=0,0
lst = []
for i in range(len(X)):
    y_pred = w*X[i] + b
    mse = (y[i] - y_pred)**2
    lst.append(mse)
mse = sum(lst)/len(lst)
print(mse)
```

*When w and b to 0 then MSE=27.666*

**EPOCHS1**

```python
g_w,g_b = -22.67,-10
learning_rate = 0.01
w_update = 0 - learning_rate*g_w
b_update = 0 - learning_rate*g_b
print(w_update,b_update)
lst = []
for i in range(len(X)):
    y_pred = w_update*X[i] + b_update
    mse = (y[i] - y_pred)**2
    lst.append(mse)
mse = sum(lst)/len(lst)
print(mse)
```

**EPOCHS2**

```python
w_update = 0.2267
b_update = 0.1
y_pred_update = []
for i in range(len(X)):
    pred = w_update*X[i] + b_update
    y_pred_update.append(pred)
g_w,g_b = 0,0
for i in range(len(X)):
    g_w += X[i]*(y_pred_update[i] - y[i])
g_w = 2*g_w/len(X)
print(g_w)
for i in range(len(X)):
    g_b += (y_pred_update[i] - y[i])
g_b = 2*g_b/len(X)
print(g_b)
w_update1 = w_update - learning_rate*g_w
b_update1 = b_update - learning_rate*g_b
lst = []
for i in range(len(X)):
    y_pred = w_update1*X[i] + b_update1
    mse = (y[i] - y_pred)**2
    lst.append(mse)
mse = sum(lst)/len(lst)
print(mse)
print(w_update1,b_update1)
```

**We can see  after each epoch the MSE is decrease**
