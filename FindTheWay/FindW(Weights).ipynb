```python
import numpy as np
X = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 6]
x_mean = np.mean(X)
y_mean = np.mean(y)
print(x_mean,y_mean)
lst = []
lst1 = []
for i in range(len(X)):
    a = (X[i]-x_mean)*(y[i]-y_mean)
    b = (X[i]-x_mean)**2
    lst.append(a)
    lst1.append(b)
print(lst)
print(lst1)
```

*FIND W*

```python
lst = np.sum(lst)
lst1 = np.sum(lst1)
print(lst,lst1)
```

```python
w = lst/lst1
print(w)
```
